# MIG Cement Demand Forecasting — Step 4: Feature Engineering

**Scope of this notebook:** turn the cleaned dataset into a modelling-ready feature set. Every
feature built here is a direct response to a specific finding from EDA — this notebook doesn't
throw a generic feature-engineering checklist at the data, it builds exactly what the evidence
said we needed:

| EDA finding | Feature response |
|---|---|
| Planned pour explains only 61% of variance | Lag and rolling-window features of past consumption, to help close the gap |
| Rain has a real but modest linear effect | Rolling rainfall exposure + a heavy-rain flag |
| Temperature has ~no linear effect, may be a threshold | A near-freezing threshold flag instead of raw temperature |
| Consistent Feb seasonal dip | Cyclical (sin/cos) month encoding |
| Behaviour drives inventory risk, not consumption | Behaviour and region as categorical/encoded features |
| Cement type barely matters | Kept only as a simple categorical, no special engineering |

**Important — avoiding lookahead leakage:** every lag/rolling feature below is computed so that
day *t*'s feature only uses information available *up to and including t-1*. Getting this wrong
(e.g. a rolling mean that includes today) is one of the most common and hardest-to-notice bugs
in time-series forecasting — it makes a model look far more accurate in testing than it will
ever be in production, because it's secretly being handed the answer.


```python
import pandas as pd
import numpy as np

pd.set_option('display.float_format', lambda x: f'{x:,.2f}')

df = pd.read_parquet("../outputs/cement_demand_clean.parquet")
df = df.sort_values(["site_id", "date"]).reset_index(drop=True)
df.shape
```

## 1. Lag Features

Past consumption is the single best available predictor of near-future consumption for most
operational time series. We build lags at 1, 7, 14, and 28 days — short-term momentum (1-day),
week-over-week (7-day), and month-over-month (14/28-day) patterns. All computed **within each
site** (`groupby('site_id')`) so one site's history never leaks into another's lag values.


```python
LAG_DAYS = [1, 7, 14, 28]

for lag in LAG_DAYS:
    df[f"consumed_lag_{lag}"] = df.groupby("site_id")["consumed_tonnes"].shift(lag)

df[["site_id", "date", "consumed_tonnes"] + [f"consumed_lag_{l}" for l in LAG_DAYS]].head(10)
```

Notice the first rows per site are `NaN` for larger lags — there's no 28-day history on a
site's 5th recorded day. We handle these at the end of the notebook (Section 6).


## 2. Rolling Aggregate Features

Rolling mean and standard deviation smooth out day-to-day noise and capture a site's recent
demand *level* and *volatility*. Critically, we `shift(1)` **before** rolling — this ensures the
window for day *t* covers days *t-7* through *t-1*, never including day *t* itself.


```python
ROLLING_WINDOWS = [7, 14, 28]

for window in ROLLING_WINDOWS:
    shifted = df.groupby("site_id")["consumed_tonnes"].shift(1)
    df[f"consumed_roll_mean_{window}"] = (
        shifted.groupby(df["site_id"]).rolling(window, min_periods=3).mean().reset_index(level=0, drop=True)
    )
    df[f"consumed_roll_std_{window}"] = (
        shifted.groupby(df["site_id"]).rolling(window, min_periods=3).std().reset_index(level=0, drop=True)
    )

df[["site_id", "date", "consumed_tonnes", "consumed_roll_mean_7", "consumed_roll_std_7"]].head(10)
```

## 3. Weather-Adjusted Features

EDA found rain has a real (if modest) negative linear effect, while temperature's effect looked
negligible *linearly* — which we hypothesised was because it matters as a threshold (near-
freezing halts pours) rather than a straight-line relationship. We engineer both ideas directly.


```python
# Rolling rainfall exposure (3-day trailing sum, lagged so it doesn't include today)
df["rain_roll_sum_3"] = (
    df.groupby("site_id")["rain_mm"].shift(1).groupby(df["site_id"]).rolling(3, min_periods=1).sum()
    .reset_index(level=0, drop=True)
)

# Heavy rain flag: above the 75th percentile of rainfall network-wide
rain_p75 = df["rain_mm"].quantile(0.75)
df["is_heavy_rain"] = df["rain_mm"] > rain_p75

# Near-freezing threshold flag, rather than relying on the raw linear correlation
df["is_near_freezing"] = df["avg_temp_c"] < 2.0

print(f"Heavy rain threshold (75th pct): {rain_p75:.1f} mm")
print(f"Share of days flagged heavy rain: {df['is_heavy_rain'].mean():.1%}")
print(f"Share of days flagged near-freezing: {df['is_near_freezing'].mean():.1%}")
```

## 4. Calendar / Seasonality Features

EDA found a real, repeating February dip. Rather than feeding month-of-year in as a raw integer
(which wrongly implies December and January are "far apart" numerically, when they're
calendar-adjacent), we use a cyclical sin/cos encoding — the standard way to represent periodic
time features so a model can learn that month 12 and month 1 are close together.


```python
df["month_sin"] = np.sin(2 * np.pi * df["month"] / 12)
df["month_cos"] = np.cos(2 * np.pi * df["month"] / 12)
df["dow_sin"] = np.sin(2 * np.pi * df["day_of_week"] / 7)
df["dow_cos"] = np.cos(2 * np.pi * df["day_of_week"] / 7)

df[["date", "month", "month_sin", "month_cos", "day_of_week", "dow_sin", "dow_cos"]].head()
```

## 5. Behaviour, Region, and Demand-Pressure Features

EDA's clearest business finding was that `behavior` drives inventory *risk* far more than it
drives consumption volume — so it needs to enter the model as a proper categorical feature. We
also build a "demand pressure" ratio: how does the planned pour compare to a site's recent
average? A ratio far above 1 signals an unusually large pour relative to that site's normal
pattern — informative regardless of the site's absolute scale.


```python
df = pd.get_dummies(df, columns=["behavior", "region"], prefix=["behavior", "region"])

# Demand pressure: planned pour vs recent rolling average (guard against divide-by-zero)
df["demand_pressure_ratio"] = df["planned_pour_tonnes"] / df["consumed_roll_mean_7"].replace(0, np.nan)

behavior_cols = [c for c in df.columns if c.startswith("behavior_")]
region_cols = [c for c in df.columns if c.startswith("region_")]
print("Behaviour dummy columns:", behavior_cols)
print("Region dummy columns:", region_cols)
df[["site_id", "planned_pour_tonnes", "consumed_roll_mean_7", "demand_pressure_ratio"]].head(10)
```

## 6. Handling the Lag-Induced Missing Values

The 28-day lag and rolling windows mean each site's first ~28 days have partial or missing
feature values — this is unavoidable (there's no history before a site's first recorded day).
We drop these rows rather than impute them, because imputing a fake "history" for a site's
earliest days risks quietly fabricating signal a model would then over-trust.


```python
before = len(df)
df_features = df.dropna(subset=[f"consumed_lag_{max(LAG_DAYS)}", f"consumed_roll_mean_{max(ROLLING_WINDOWS)}"]).copy()
after = len(df_features)

print(f"Rows before dropping lag warm-up period: {before}")
print(f"Rows after: {after}  ({before - after} dropped, {(before - after) / before:.1%})")
print(f"Remaining date range: {df_features['date'].min().date()} to {df_features['date'].max().date()}")
```

Losing under 3% of rows to the warm-up period, entirely from the first month of the
dataset, is a reasonable trade for having no fabricated values feeding the model.


## 7. Final Feature Set Check

A quick sanity check before saving: confirm no unexpected nulls remain, and preview the full
feature list.


```python
remaining_nulls = df_features.isna().sum()
remaining_nulls = remaining_nulls[remaining_nulls > 0]
print("Columns with remaining nulls:")
print(remaining_nulls if len(remaining_nulls) else "None")
print()
print(f"Final shape: {df_features.shape}")
print()
print("Feature columns:")
print([c for c in df_features.columns if c not in
       ["date", "site_id", "cement_type", "consumed_tonnes"]])
```

`demand_pressure_ratio` may still carry a handful of nulls/infs on days where the rolling
mean was zero — worth capping or imputing at model-fit time in Step 5 rather than here, since
the right treatment (drop vs. cap vs. median-fill) can depend on which model is being trained.


## 8. Persist the Feature Set


```python
import os
os.makedirs("../outputs", exist_ok=True)

df_features.to_parquet("../outputs/cement_demand_features.parquet", index=False)
print(f"Saved cement_demand_features.parquet — {df_features.shape[0]} rows, {df_features.shape[1]} columns")
```
