```python
# Display the distribution of important categorical variables
# as both counts and percentages.

categorical_columns = [
    "Status",
    "PriorityID",
    "CategoryID",
    "Channel",
    "SLABreached"
]

for column in categorical_columns:
    
    # Count the number of records in each category.
    counts = tickets[column].value_counts(dropna=False)
    
    # Calculate the percentage represented by each category.
    percentages = (
        tickets[column]
        .value_counts(normalize=True, dropna=False)
        .mul(100)
        .round(2)
    )
    
    # Combine counts and percentages into one table.
    distribution = pd.DataFrame({
        "Count": counts,
        "Percentage": percentages
    })
    
    print(f"\n{column} Distribution:")
    display(distribution)
```

```python
# List the date columns that need to be checked.
date_columns = [
    "CreatedAt",
    "FirstResponseAt",
    "ResolvedAt",
    "SLADueAt"
]

# Profile each date column without modifying the original dataframe.
for column in date_columns:
    
    # Convert the column temporarily to datetime.
    # Invalid date values will become NaT (Not a Time).
    converted_dates = pd.to_datetime(
        tickets[column],
        errors="coerce"
    )
    
    # Count values that could not be converted,
    # excluding values that were already missing.
    invalid_dates = (
        converted_dates.isna() & tickets[column].notna()
    ).sum()
    
    print(f"\n{column}:")
    print("Invalid dates:", invalid_dates)
    print("Earliest date:", converted_dates.min())
    print("Latest date:", converted_dates.max())
```

```python
# Temporarily convert the date columns to datetime
# so that we can check the logical order of events.
created = pd.to_datetime(tickets["CreatedAt"], errors="coerce")
first_response = pd.to_datetime(tickets["FirstResponseAt"], errors="coerce")
resolved = pd.to_datetime(tickets["ResolvedAt"], errors="coerce")
sla_due = pd.to_datetime(tickets["SLADueAt"], errors="coerce")

# Check for impossible or suspicious date relationships.
response_before_creation = (
    first_response.notna() & (first_response < created)
).sum()

resolution_before_creation = (
    resolved.notna() & (resolved < created)
).sum()

sla_before_creation = (
    sla_due.notna() & (sla_due < created)
).sum()

print("First response before ticket creation:", response_before_creation)
print("Resolution before ticket creation:", resolution_before_creation)
print("SLA due before ticket creation:", sla_before_creation)
```

```python
# Join client contract information to the ticket data
# for profiling purposes only.

tickets_clients = tickets.merge(
    clients[
        ["ClientID", "ContractTier", "SLACreditClause"]
    ],
    on="ClientID",
    how="left"
)

# Confirm that the join did not change the number of tickets.
print("Tickets before join:", len(tickets))
print("Tickets after join:", len(tickets_clients))
```

```python
# Calculate SLA breach rate by Contract Tier.
# The mean of a binary 0/1 variable gives the proportion of breached tickets.

breach_by_contract = (
    tickets_clients
    .groupby("ContractTier")["SLABreached"]
    .agg(["count", "sum", "mean"])
    .rename(columns={
        "count": "Total_Tickets",
        "sum": "Breached_Tickets",
        "mean": "Breach_Rate"
    })
)

# Convert the breach rate to a percentage.
breach_by_contract["Breach_Rate"] = (
    breach_by_contract["Breach_Rate"] * 100
).round(2)

display(breach_by_contract)
```

```python
# Calculate SLA breach rate based on whether
# the client has an SLA credit clause.

breach_by_credit_clause = (
    tickets_clients
    .groupby("SLACreditClause")["SLABreached"]
    .agg(["count", "sum", "mean"])
    .rename(columns={
        "count": "Total_Tickets",
        "sum": "Breached_Tickets",
        "mean": "Breach_Rate"
    })
)

# Convert the breach rate to a percentage.
breach_by_credit_clause["Breach_Rate"] = (
    breach_by_credit_clause["Breach_Rate"] * 100
).round(2)

display(breach_by_credit_clause)
```

### SLA Breach by Client Characteristics
